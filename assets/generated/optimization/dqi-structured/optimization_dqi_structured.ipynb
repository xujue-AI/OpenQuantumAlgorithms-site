# Degree-one structured DQI: saved-artifact companion

This ordinary Python notebook reads the complete portable bundle. It verifies saved files and scalar references, displays the new clean-workspace circuit, and separates the small coherent checks from the larger generation-only example. It does not run a simulator or invoke circuit generation.

The source packet alone is not a validated result. A complete bundle needs the fresh n=2 and n=3 validation artifacts and the n=8 generation artifacts produced by the public commands. Missing or source-only data fail clearly; analytical predictions are not substituted for executed evidence.

The family is B=[I_n;1_n^T], degree one, with signs (0,...,0,1). Workspace starts clean. The original default's stronger arbitrary-dirty decoder and 377/396 counts remain a separate baseline.

````python
from pathlib import Path
import hashlib
import json
import types
import sys
from IPython.display import Markdown, display

candidates = [Path.cwd(), Path.cwd() / 'assets/generated/optimization/dqi-structured', Path.cwd().parent / 'assets/generated/optimization/dqi-structured']
bundle = next((path.resolve() for path in candidates if (path / 'bundle.json').is_file() and (path / 'verify_dqi_structured_saved.py').is_file()), None)
if bundle is None:
    raise FileNotFoundError('Keep the complete structured DQI bundle together, or run from its repository notebook directory.')
reader_path = bundle / 'verify_dqi_structured_saved.py'
with reader_path.open('rb') as stream:
    reader_source = stream.read((1 << 20) + 1)
if len(reader_source) > (1 << 20):
    raise ValueError('Saved reader source exceeds its fixed byte cap.')
with (bundle / 'bundle.json').open('rb') as stream:
    manifest_source = stream.read((1 << 20) + 1)
if len(manifest_source) > (1 << 20):
    raise ValueError('Bundle manifest exceeds its fixed byte cap.')
reader_binding = json.loads(manifest_source)['files'].get('verify_dqi_structured_saved.py')
if reader_binding != {'kind': 'file', 'bytes': len(reader_source), 'sha256': hashlib.sha256(reader_source).hexdigest()}:
    raise ValueError('The reader buffer differs from the portable manifest binding.')
reader = types.ModuleType('openquantum_structured_saved_reader')
reader.__file__ = str(reader_path)
sys.modules[reader.__name__] = reader
exec(compile(reader_source, str(reader_path), 'exec'), reader.__dict__)
verified = reader.verify_saved_bundle(bundle)
report = verified['verification']
print('Saved n=2/3 clean-workspace coherent checks; n=8 generation-only. This notebook repeats no quantum simulation.')
````
