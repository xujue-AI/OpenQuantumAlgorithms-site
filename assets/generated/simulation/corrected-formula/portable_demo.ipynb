```python
from pathlib import Path
import importlib.util, sys
packet = next(path for path in (Path.cwd(), *(parent / "assets/generated/simulation/corrected-formula" for parent in (Path.cwd(), *Path.cwd().parents))) if (path / "generate_cpf2_public.py").is_file())
spec = importlib.util.spec_from_file_location("cpf2_public_wrapper", packet / "generate_cpf2_public.py")
wrapper = importlib.util.module_from_spec(spec)
sys.modules[spec.name] = wrapper
spec.loader.exec_module(wrapper)
cpf, packet, original_sources = wrapper.load_science(packet)
```

The declared known input uses H on both wires and S on the lower wire. Every complete trial includes resets, both corrector boundaries, all Strang steps and both readout bits.


```python
model = cpf.PairModel(1, alpha="1/5")
circuit = cpf.complete_trial(model, "1", 4)
circuit.draw("mpl", fold=40, cregbundle=False)
```
