```python
from pathlib import Path
from fractions import Fraction
import hashlib, importlib.util, json, sys
from IPython.display import SVG, Markdown, display

candidates = [Path.cwd(), *Path.cwd().parents]
ROOT = next((p for p in candidates if (p / 'assets/generated/optimization/word-transport/record-manifest.json').is_file() and (p / 'scripts/optimization_word_record_release.py').is_file()), None)
if ROOT is None:
    raise RuntimeError('Run this notebook inside the complete extracted examples package')
helper = ROOT / 'scripts/optimization_word_record_release.py'
raw = helper.read_bytes()
assert hashlib.sha256(raw).hexdigest() == '2674bfe0dab04d4452b69ed740ae3e3f4baebec7ed988407a42e4c6e40d339d7'
spec = importlib.util.spec_from_file_location('notebook_word_record_release', helper)
release = importlib.util.module_from_spec(spec)
sys.modules[spec.name] = release
exec(compile(raw, str(helper), 'exec'), release.__dict__)
BUNDLE = ROOT / 'assets/generated/optimization/word-transport'
verified = release.verify_public_word_transport(ROOT)
assert verified['native_api_calls'] == 0 and verified['writes_performed'] is False
print({key: verified[key] for key in ('status','transcript_id','classes','parents','groups','preparations','t_states')})
```
