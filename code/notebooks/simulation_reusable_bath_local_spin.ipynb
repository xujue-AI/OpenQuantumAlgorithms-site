```python
from pathlib import Path, PurePosixPath
from hashlib import sha256
import json
from IPython.display import JSON, display

MAX_FILES = 32
MAX_TOTAL_BYTES = 32 * 1024 * 1024
MAX_MANIFEST_BYTES = 512 * 1024
MAX_QASM_BYTES = 8 * 1024 * 1024
EXPECTED_CASE_SPEC_SHA256 = "761c2ca3eafabf672a221d18abfdfbae91b024f25ee3a41df4b5c95de1d041f3"
EXPECTED_KIND = "local-spin-bath-saved-evidence"
EXPECTED_FILES = frozenset({
    "cases.json", "generation.json", "schedules.json", "reference.json", "validation.json",
    "source-exports/n1-ground.qasm", "source-exports/n1-thermal.qasm",
    "source-exports/n2-ground.qasm", "source-exports/n2-thermal.qasm",
    "source-exports/n4-thermal-generation.qasm",
    "source-exports/n8-thermal-generation.qasm",
    "source-exports/n16-thermal-generation.qasm",
    "source-exports/n32-thermal-generation.qasm",
    "source-exports/n64-thermal-generation.qasm",
    "source/reusable_bath_local_spin.py",
    "source/reusable_bath_local_spin_reference.py",
    "source/cases.json", "source/requirements-lock.txt", "source/LICENSE.txt",
})
MANIFEST_NAME = "artifact-manifest.json"
SIDECAR_NAME = "artifact-manifest.sha256"

def unique_json_object(pairs):
    result = {}
    for key, value in pairs:
        if key in result:
            raise ValueError("Duplicate key in saved JSON: " + key)
        result[key] = value
    return result

def reject_json_constant(value):
    raise ValueError("Nonfinite JSON constant in saved evidence: " + value)

def decode_saved_json(data):
    return json.loads(
        data.decode("utf-8"),
        object_pairs_hook=unique_json_object,
        parse_constant=reject_json_constant,
    )

def valid_sha256(value):
    return (type(value) is str and len(value) == 64
            and all(character in "0123456789abcdef" for character in value))

here = Path.cwd()
candidates = (
    here,
    here / "reusable-bath-local-spin-evidence",
    here / "assets/generated/simulation/reusable-bath-local-spin",
    here.parent / "assets/generated/simulation/reusable-bath-local-spin",
)
artifact_root = None
for candidate in candidates:
    if (not candidate.is_symlink() and candidate.is_dir()
            and (candidate / MANIFEST_NAME).exists()):
        artifact_root = candidate
        break

saved_files = {}
saved_objects = {}
manifest = {}
evidence_loaded = False

def read_regular_saved_file(name, expected_bytes=None, byte_limit=MAX_TOTAL_BYTES):
    relative = PurePosixPath(name)
    if (type(name) is not str or not name or relative.is_absolute()
            or ".." in relative.parts or relative.as_posix() != name
            or "\\" in name):
        raise ValueError("Saved filename is not a portable relative path")
    current = artifact_root
    for part in relative.parts[:-1]:
        current = current / part
        if current.is_symlink() or not current.is_dir():
            raise ValueError("Saved directory is nonregular or symlinked")
    target = artifact_root / relative
    if target.is_symlink() or not target.is_file():
        raise ValueError("Saved artifact is nonregular or symlinked: " + name)
    size = target.lstat().st_size
    if size < 0 or size > byte_limit or (expected_bytes is not None and size != expected_bytes):
        raise ValueError("Saved artifact length is outside its bound: " + name)
    with target.open("rb") as saved_input:
        data = saved_input.read(size + 1)
    if len(data) != size:
        raise ValueError("Saved artifact length changed while reading: " + name)
    return data

if artifact_root is None:
    print("SOURCE PREPARED; VALIDATION PLANNED.")
    print("Saved local-spin evidence is currently unavailable; no scientific work was run.")
else:
    manifest_bytes = read_regular_saved_file(MANIFEST_NAME, byte_limit=MAX_MANIFEST_BYTES)
    sidecar_bytes = read_regular_saved_file(SIDECAR_NAME, expected_bytes=65, byte_limit=65)
    sidecar_text = sidecar_bytes.decode("ascii")
    if (not sidecar_text.endswith("\n") or not valid_sha256(sidecar_text[:-1])
            or sha256(manifest_bytes).hexdigest() != sidecar_text[:-1]):
        raise ValueError("Public manifest sidecar integrity mismatch")
    manifest = decode_saved_json(manifest_bytes)
    if (type(manifest) is not dict or type(manifest.get("schema_version")) is not int
            or manifest.get("schema_version") != 1
            or manifest.get("kind") != EXPECTED_KIND
            or manifest.get("all_components_verified") is not False):
        raise ValueError("Unexpected saved manifest kind or scientific scope")
    records = manifest.get("files")
    if type(records) is not dict or set(records) != EXPECTED_FILES:
        raise ValueError("Manifest payloads do not match the public packet")
    if len(records) + 2 > MAX_FILES:
        raise ValueError("Saved packet exceeds the file limit")

    total_bytes = len(manifest_bytes) + len(sidecar_bytes)
    for name, record in records.items():
        if (type(record) is not dict or set(record) != {"bytes", "sha256"}
                or type(record["bytes"]) is not int or record["bytes"] < 0
                or not valid_sha256(record["sha256"])):
            raise ValueError("Malformed saved file record: " + name)
        if name.endswith(".qasm") and record["bytes"] > MAX_QASM_BYTES:
            raise ValueError("Saved QASM exceeds its individual byte limit")
        total_bytes += record["bytes"]
        if total_bytes > MAX_TOTAL_BYTES:
            raise ValueError("Saved packet exceeds 32 MiB")

    actual_files = set()
    for child in artifact_root.iterdir():
        if child.is_symlink():
            raise ValueError("Saved packet contains a symlink")
        if child.is_file():
            actual_files.add(child.name)
        elif child.name in {"source", "source-exports"} and child.is_dir():
            for entry in child.iterdir():
                if entry.is_symlink() or not entry.is_file():
                    raise ValueError("Saved payload directory contains a nonregular file")
                actual_files.add(child.name + "/" + entry.name)
                if len(actual_files) > MAX_FILES:
                    raise ValueError("Saved packet exceeds the file limit")
        else:
            raise ValueError("Unexpected directory or nonregular saved artifact")
        if len(actual_files) > MAX_FILES:
            raise ValueError("Saved packet exceeds the file limit")
    if actual_files != EXPECTED_FILES | {MANIFEST_NAME, SIDECAR_NAME}:
        raise ValueError("Public saved packet contains missing or extra files")

    for name in sorted(EXPECTED_FILES):
        record = records[name]
        data = read_regular_saved_file(name, expected_bytes=record["bytes"])
        if sha256(data).hexdigest() != record["sha256"]:
            raise ValueError("Saved artifact integrity mismatch: " + name)
        saved_files[name] = data
    for name in ("cases.json", "source/cases.json"):
        if sha256(saved_files[name]).hexdigest() != EXPECTED_CASE_SPEC_SHA256:
            raise ValueError("Saved scientific case specification changed")
    for name in ("cases.json", "generation.json", "schedules.json", "reference.json", "validation.json"):
        value = decode_saved_json(saved_files[name])
        if type(value) is not dict:
            raise ValueError("Saved report must be a JSON object: " + name)
        saved_objects[name] = value
    for name in ("reference.json", "validation.json"):
        if saved_objects[name].get("all_components_verified") is not False:
            raise ValueError("Saved finite-component scope changed: " + name)
    if type(saved_objects["validation.json"].get("status")) is not str:
        raise ValueError("Saved validation status is missing")
    evidence_loaded = True
    print("Verified integrity of", len(actual_files), "regular saved files;", total_bytes, "bytes.")
    print("No scientific source or numerical reference was executed by the loader.")

def show_saved(name, value):
    # Wrap every value in an object; IPython JSON also accepts scalar report fields this way.
    display(JSON({name: value}, expanded=False))
```
