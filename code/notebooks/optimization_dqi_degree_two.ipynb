````python
from pathlib import Path
import hashlib
import json
import sys
import types
from IPython.display import Markdown, display

READER_SHA256 = "803ea2b04d1d5c141361cdebb98f7996b973e268625df340210d1d61042fed56"
candidates = (
    Path.cwd(),
    Path.cwd() / "assets/generated/optimization/dqi-degree-two",
    Path.cwd().parent / "assets/generated/optimization/dqi-degree-two",
)
bundle = next((path for path in candidates if (path / "bundle.json").is_file()
               and (path / "verify_dqi_degree_two_saved.py").is_file()), None)
if bundle is None:
    raise FileNotFoundError("Keep the complete degree-two saved bundle together, or run from its repository notebook directory.")
reader_path = bundle / "verify_dqi_degree_two_saved.py"
if reader_path.is_symlink():
    raise ValueError("The saved reader must be an ordinary file.")
with reader_path.open("rb") as stream:
    reader_source = stream.read((128 << 10) + 1)
if len(reader_source) > (128 << 10) or hashlib.sha256(reader_source).hexdigest() != READER_SHA256:
    raise ValueError("The pure saved-reader source differs from this notebook's frozen binding.")
reader = types.ModuleType("degree_two_saved_reader")
reader.__file__ = str(reader_path)
exec(compile(reader_source, str(reader_path), "exec"), reader.__dict__)
verified = reader.verify_saved_bundle(bundle)
report = verified["validation"]
print("Saved C5 checks and a figure-eight source export are authenticated. This notebook repeats no quantum work.")
print("Notebook source/type projection:", verified["notebook_source_projection_sha256"])
````
